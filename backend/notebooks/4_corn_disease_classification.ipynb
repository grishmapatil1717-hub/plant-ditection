# 🌿 Corn Disease Detection using Deep Learning
**Plant:** `Corn`  
**Target Classes:** `Cercospora Leaf Spot, Common Rust, Northern Leaf Blight, Healthy`  
**Pipeline:** Transfer Learning / Custom CNN with Data Augmentation & Evaluation

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import layers, models

print('TensorFlow Version:', tf.__version__)
print('GPU Available:', len(tf.config.list_physical_devices('GPU')) > 0)
```

```python
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
CLASSES = ['Cercospora Leaf Spot', 'Common Rust', 'Northern Leaf Blight', 'Healthy']

# Data Generators with Augmentation
datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=20,
    zoom_range=0.15,
    horizontal_flip=True,
    validation_split=0.2
)
```

```python
# Build MobileNetV2 Architecture
base = tf.keras.applications.MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')
base.trainable = False

model = models.Sequential([
    base,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(len(CLASSES), activation='softmax')
])

model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
# Save Trained Model
model.save('corn_model.keras')
print('Model successfully saved to corn_model.keras')
```
